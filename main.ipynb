# Text preprocessing

```python
data['product_name'] = data['product_name'].str.lower().str.strip()
```

```python
data['combined_features'] = data['product_name'] + " " + data['category'].astype(str)
```

```python
data.head()
```

# TF-IDF Vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(data['combined_features'])

print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

# Calculating similarity scores

```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

print("Cosine similarity matrix shape:", cosine_sim.shape)
```

# Function to get similar products

```python
def get_recommendations(product_name, cosine_sim=cosine_sim):
    idx = data[data['product_name'].str.lower() == product_name.lower()].index

    if len(idx) == 0:
        return "Sorry, product not found in the dataset."
    
    idx = idx[0]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:6]  # top 5 similar products
    product_indices = [i[0] for i in sim_scores]
    
    return data['product_name'].iloc[product_indices].values
```

# Evaluate Recommendation Quality

```python
def precision_at_k(actual, predicted, k=5):
    predicted = list(predicted)[:k]
    actual = list(actual)
    if k == 0:
        return 0.0
    return len(set(predicted) & set(actual)) / float(k)

def recall_at_k(actual, predicted, k=5):
    predicted = list(predicted)[:k]
    actual = list(actual)
    if len(actual) == 0:
        return 0.0
    return len(set(predicted) & set(actual)) / float(len(actual))

def recommend_for_user(user_id, num_recommendations=5):
    """
    Simple user-level recommender that:
    - finds products the user has interacted with (df['user_id'] may contain comma-separated ids)
    - for each such product, gets content-based recommendations via get_recommendations (by product_name)
    - maps recommended product names back to product_ids (from `data`) and returns top N unique ids
    """
    # find product_ids the user has interacted with
    mask = df['user_id'].astype(str).str.contains(str(user_id))
    user_products = df.loc[mask, 'product_id'].unique().tolist()
    if len(user_products) == 0:
        return []

    rec_ids = []
    for pid in user_products:
        # find product_name in `data`
        rows = data[data['product_id'] == pid]
        if rows.empty:
            continue
        pname = rows['product_name'].iloc[0]
        rec_names = get_recommendations(pname)  # returns product_name strings
        if isinstance(rec_names, str):
            # get_recommendations returns an error message string when not found
            continue
        # map recommended names to product_ids
        ids = data.loc[data['product_name'].isin(rec_names), 'product_id'].tolist()
        for i in ids:
            if i not in user_products and i not in rec_ids:
                rec_ids.append(i)
            if len(rec_ids) >= num_recommendations:
                break
        if len(rec_ids) >= num_recommendations:
            break

    return rec_ids[:num_recommendations]

# Example usage:
sample_user = df['user_id'].iloc[0]            # use correct column name
actual_items = df[df['user_id'].astype(str).str.contains(str(sample_user))]['product_id'].unique().tolist()
predicted_items = recommend_for_user(sample_user, num_recommendations=5)

print("Actual items ({}):".format(len(actual_items)), actual_items)
print("Predicted items:", predicted_items)
print("Precision:", precision_at_k(actual_items, predicted_items))
print("Recall:", recall_at_k(actual_items, predicted_items))
```
